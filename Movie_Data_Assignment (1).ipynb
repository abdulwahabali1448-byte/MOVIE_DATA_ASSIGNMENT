### **Question 16**

### **Question 29**
Har movie ke 'Genre' column ko split karke check karein ki total unique genres pure dataset mein kitne hain.

```python
import pandas as pd

df = pd.read_csv('movie.csv')

genre_col = None
for col in ['genre', 'Genre', 'genres', 'Genres']:
    if col in df.columns:
        genre_col = col
        break

if genre_col:
    df[genre_col] = df[genre_col].astype(str)
    
    # Comma (,) aur Pipe (|) dono cases ko handle karne ke liye split kar rahe hain
    all_genres = df[genre_col].str.split(r'[\s]*,[\s]*|[\s]*\|[\s]*')
    
    # Saare lists ka ek single flat combination bana rahe hain
    flat_genres = [genre.strip() for sublist in all_genres.dropna() for genre in sublist if genre.strip() != 'nan']
    
    unique_genres = sorted(list(set(flat_genres)))
    
    print(f"Total Unique Genres: {len(unique_genres)}")
    print("\nList of Unique Genres:")
    print(unique_genres)
else:
    print("Genre column nahi mila. Aapke dataset ke columns yeh hain:")
    print(df.columns)
```

### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
import pandas as pd

df = pd.read_csv('movie.csv')

# --- 1. Runtime Column Cleaning ---
runtime_col = next((c for c in ['runtime', 'Runtime', 'duration', 'Duration'] if c in df.columns), None)
if runtime_col:
    df[runtime_col] = df[runtime_col].astype(str).str.replace(' min', '', case=False).str.strip()
    df[runtime_col] = pd.to_numeric(df[runtime_col], errors='coerce').astype('Int64')

# --- 2. Gross Column Cleaning & Imputation ---
gross_col = next((c for c in ['Gross', 'gross', 'Gross_Earnings', 'earnings'] if c in df.columns), None)
if gross_col:
    df[gross_col] = df[gross_col].astype(str).str.replace(',', '').str.replace('$', '').str.strip()
    df[gross_col] = pd.to_numeric(df[gross_col], errors='coerce')
    median_value = df[gross_col].median()
    df[gross_col] = df[gross_col].fillna(median_value)

# --- 3. Rating Columns Numeric Conversion ---
rating_col = next((c for c in ['IMDB_Rating', 'imdb_rating', 'rating', 'score'] if c in df.columns), None)
if rating_col:
    df[rating_col] = pd.to_numeric(df[rating_col], errors='coerce')
    df['Rating_100_Scale'] = df[rating_col] * 10

meta_col = next((c for c in ['Meta_score', 'meta_score', 'Metascore'] if c in df.columns), None)
if meta_col:
    df[meta_col] = pd.to_numeric(df[meta_col], errors='coerce')

# --- 4. Safely Saving the Cleaned Dataset ---
df.to_csv('cleaned_movies.csv', index=False)
print("Dataset successfully cleaned and saved as 'cleaned_movies.csv'!")
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('movie.csv')

genre_col = next((c for c in ['Genre', 'genre', 'genres'] if c in df.columns), None)
gross_col = next((c for c in ['Gross', 'gross', 'Gross_Earnings', 'earnings'] if c in df.columns), None)
title_col = next((c for c in ['Series_Title', 'title', 'Title'] if c in df.columns), None)

if genre_col and gross_col:
    # 1. Gross column ko clean aur numeric mein convert karein
    df[gross_col] = df[gross_col].astype(str).str.replace(',', '').str.replace('$', '').str.strip()
    df[gross_col] = pd.to_numeric(df[gross_col], errors='coerce')
    
    # 2. Drama aur Action ki separate rows filter karke combine karein
    df[genre_col] = df[genre_col].astype(str)
    
    action_movies = df[df[genre_col].str.contains('Action', case=False, na=False)].copy()
    action_movies['Target_Genre'] = 'Action'
    
    drama_movies = df[df[genre_col].str.contains('Drama', case=False, na=False)].copy()
    drama_movies['Target_Genre'] = 'Drama'
    
    # Dono filtered subsets ka data combine karein
    plot_data = pd.concat([action_movies, drama_movies]).dropna(subset=[gross_col])
    
    # Millions mein convert karne ke liye (agar revenue large numbers mein ho)
    if plot_data[gross_col].max() > 1000000:
        plot_data['Gross_Millions'] = plot_data[gross_col] / 1000000
        y_axis_col = 'Gross_Millions'
        y_label = 'Gross Revenue (in Millions $)'
    else:
        y_axis_col = gross_col
        y_label = 'Gross Revenue'

    # 3. Violin Plot banayein
    plt.figure(figsize=(10, 6))
    sns.violinplot(
        data=plot_data, 
        x='Target_Genre', 
        y=y_axis_col, 
        palette='muted', 
        inner='quartile'
    )
    
    plt.title('Gross Revenue Comparison: Action vs Drama Movies', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Genre', fontsize=12, fontweight='bold')
    plt.ylabel(y_label, fontsize=12, fontweight='bold')
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    
    plt.tight_layout()
    plt.show()
else:
    print("Zaroori columns nahi mile. Aapke dataset ke columns yeh hain:")
    print(df.columns)
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('movie.csv')

rating_col = next((c for c in ['IMDB_Rating', 'imdb_rating', 'rating', 'score'] if c in df.columns), None)
meta_col = next((c for c in ['Meta_score', 'meta_score', 'Metascore', 'metascore'] if c in df.columns), None)
title_col = next((c for c in ['Series_Title', 'title', 'Title', 'movie_title'] if c in df.columns), None)

if all([rating_col, meta_col, title_col]):
    df[rating_col] = pd.to_numeric(df[rating_col], errors='coerce')
    df[meta_col] = pd.to_numeric(df[meta_col], errors='coerce')
    
    # 1. Extract Top 5 highest rated movies (drop NaNs in ratings/metascores to ensure clean comparison)
    top_5 = df.dropna(subset=[rating_col, meta_col]).sort_values(by=rating_col, ascending=False).head(5).copy()
    
    # 2. Scale IMDB Rating to 100
    top_5['IMDB_Rating_100'] = top_5[rating_col] * 10
    
    # 3. Melt the dataframe to make it suitable for a grouped side-by-side seaborn barplot
    melted_df = pd.melt(
        top_5, 
        id_vars=[title_col], 
        value_vars=['IMDB_Rating_100', meta_col],
        var_name='Rating_Type', 
        value_name='Score'
    )
    
    # Clean up labels for the legend
    melted_df['Rating_Type'] = melted_df['Rating_Type'].replace({'IMDB_Rating_100': 'IMDB Rating (x10)', meta_col: 'Meta Score'})
    
    # 4. Create the plot
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(
        data=melted_df, 
        x=title_col, 
        y='Score', 
        hue='Rating_Type', 
        palette='Set2',
        edgecolor='black'
    )
    
    plt.title('Top 5 Highest Rated Movies: IMDB Rating vs Meta Score', fontsize=15, fontweight='bold', pad=15)
    plt.xlabel('Movie Title', fontsize=12, fontweight='bold')
    plt.ylabel('Scores (Out of 100)', fontsize=12, fontweight='bold')
    plt.ylim(0, 115)  # Leave headroom for text labels
    plt.xticks(rotation=15, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    
    # Add values on top of the bars
    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(
                f'{p.get_height():.1f}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', 
                va='center', 
                xytext=(0, 8), 
                textcoords='offset points',
                fontsize=10, 
                fontweight='bold'
            )
            
    plt.legend(title='Score Type', loc='upper right')
    plt.tight_layout()
    plt.show()
else:
    print("Required columns missing. Your dataset columns are:")
    print(df.columns)
```
